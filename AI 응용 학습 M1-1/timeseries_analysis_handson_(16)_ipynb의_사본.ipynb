````python
# ▶️ 실행 하세요 (1분 정도 소요 예상)
!pip install -q yfinance holidays
!apt-get -qq install -y fonts-nanum > /dev/null

import os, json, zipfile, warnings
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

warnings.filterwarnings("ignore")

# 한글 폰트 등록
# 주의: 방금 설치한 폰트는 matplotlib 캐시에 없으므로 파일을 직접 addfont 해야 인식됩니다.
def setup_korean_font():
    candidates = [
        "/usr/share/fonts/truetype/nanum/NanumGothic.ttf",
        "/usr/share/fonts/truetype/nanum/NanumBarunGothic.ttf",
    ]
    candidates += [f for f in fm.findSystemFonts() if "Nanum" in os.path.basename(f)]
    for path in candidates:
        if os.path.exists(path):
            try:
                fm.fontManager.addfont(path)
                name = fm.FontProperties(fname=path).get_name()
                plt.rcParams["font.family"] = name
                return name
            except Exception:
                continue
    return None

KOREAN_FONT = setup_korean_font()
plt.rcParams["axes.unicode_minus"] = False
plt.rcParams["figure.figsize"] = (12, 5)
plt.rcParams["axes.grid"] = True
plt.rcParams["grid.alpha"] = 0.3

print("pandas :", pd.__version__)
print("numpy  :", np.__version__)
print("한글 폰트 :", KOREAN_FONT or "❌ 실패")

# 한글이 실제로 렌더링되는지 검증 (누락 글리프가 있으면 경고가 발생)
import warnings as _w
with _w.catch_warnings(record=True) as _rec:
    _w.simplefilter("always")
    _f, _a = plt.subplots(figsize=(1, 1))
    _a.set_title("한글")
    _f.canvas.draw()
    plt.close(_f)
_broken = any("missing from font" in str(x.message) for x in _rec)

if KOREAN_FONT and not _broken:
    print("✅ 환경 준비 완료 — 그래프 한글 정상")
else:
    print("⚠️ 한글이 □로 깨질 수 있습니다.")
    print("   해결: 상단 메뉴 [런타임 > 세션 다시 시작] 후 이 셀부터 다시 실행하세요.")
````

### 데이터 불러오기 ▶️

인터넷에서 바로 받아옵니다. 몇 초 걸립니다.

> 🔌 네트워크 사정으로 가끔 실패할 수 있어 **자동으로 3번까지 재시도**합니다. 그래도 안 되면 아래 안내를 따라 주세요.

````python
# ▶️ 실행 하세요
import time


def load_yfinance(ticker, start, end, retry=3):
    """Yahoo Finance에서 일별 종가를 받아 date / value 두 컬럼으로 정리한다."""
    import yfinance as yf

    last_err = None
    for attempt in range(1, retry + 1):
        try:
            raw = yf.download(ticker, start=start, end=end,
                              progress=False, auto_adjust=True)
            if raw is None or len(raw) == 0:
                raise ValueError("받아온 데이터가 비어 있습니다.")

            if isinstance(raw.columns, pd.MultiIndex):
                raw.columns = raw.columns.get_level_values(0)

            out = pd.DataFrame({"value": raw["Close"].astype(float)})  # 환율도 Close 컬럼으로 제공됨
            out.index = pd.to_datetime(out.index)
            out.index.name = "date"
            return out.sort_index()

        except Exception as e:
            last_err = e
            if attempt < retry:
                print(f"   ...{attempt}번째 시도 실패, 재시도합니다 ({e})")
                time.sleep(2)

    raise RuntimeError(f"{retry}번 시도했지만 데이터를 받지 못했습니다. 원인: {last_err}")


print(f"📡 {TICKER} 데이터를 받아오는 중... ({START_DATE} ~ {END_DATE})")

try:
    df_raw = load_yfinance(TICKER, START_DATE, END_DATE)
    print(f"\n✅ 데이터 {len(df_raw)}행 로드 완료 "
          f"({df_raw.index.min().date()} ~ {df_raw.index.max().date()})")
    display(df_raw.head())

except Exception as e:
    print("\n❌ 데이터를 받지 못했습니다.")
    print(f"   {e}\n")
    print("   이 순서로 확인해 보세요.")
    print("   1. 이 셀을 한 번 더 실행합니다 (일시적인 네트워크 문제가 대부분입니다).")
    print("   2. TICKER 철자를 확인합니다. 환율은 KRW=X 처럼 뒤에 '=X'가 붙습니다.")
    print("   3. START_DATE < END_DATE 인지, 미래 날짜가 아닌지 확인합니다.")
    print("   4. STEP 0을 다시 실행해 yfinance가 설치됐는지 확인합니다.")
    print("   5. 그래도 안 되면 [런타임 > 세션 다시 시작] 후 STEP 0부터 다시 실행합니다.")
    raise
````

# STEP 2-2. 거래일 정합성 검증 ▶️ **(데이터를 믿기 전에 확인할 것)**

`결측치 0건`이 나왔다고 안심하면 안 됩니다. **"빠진 값이 없다"와 "올바른 날짜만 들어 있다"는 다른 문제**입니다.

Yahoo Finance의 `KRW=X`는 서울외국환중개가 고시하는 국내 환율이 아니라 **글로벌 장외(OTC) 호가**입니다.
그래서 서울 외환시장이 쉬는 날에도 값이 채워져 들어옵니다. 실제로 확인된 사례:

| 확인 항목 | 결과 |
|---|---|
| 2024-01-01(신정, 서울 휴장) | Yahoo에는 값이 존재. 직전 서울 종가(1,288.0원)와 5.5원 차이 |
| 2025-04-09 종가 | 서울 주간종가 1,484.10원 / Yahoo 1,486.13원 → **2.03원 차이** |
| 같은 계열 데이터의 일요일 행 | 존재함 (어떤 시장도 열지 않는 날) |
| 최고·최저·급변동일의 **날짜** | 서울 기록과 **일치**. 날짜가 하루씩 밀리지는 않음 |

즉 날짜가 밀리는 문제는 아니고, **거래일 집합과 종가 시점이 다른** 문제입니다.
아래 셀에서 직접 확인한 뒤, 서울 거래일 기준으로 데이터를 정렬합니다.

````python
# ▶️ 실행 — 진단: 지금 데이터에 서울 휴장일이 섞여 있는가
from datetime import date, timedelta
import holidays as pyholidays

def kr_market_holidays(years):
    # 공휴일 + 근로자의날(5/1) + 연말 휴장일(12/31). 금융시장은 이 날 모두 쉰다.
    hs = set(pyholidays.KR(years=list(years)).keys())
    for y in years:
        hs.add(date(y, 5, 1))
        hs.add(date(y, 12, 31))
    return hs

YEARS = range(df_raw.index.min().year, df_raw.index.max().year + 1)
KR_HOLIDAYS = kr_market_holidays(YEARS)

idx_dates = pd.Series(df_raw.index.date, index=df_raw.index)
is_weekend = pd.Series(df_raw.index.dayofweek >= 5, index=df_raw.index)
is_holiday = idx_dates.isin(KR_HOLIDAYS)

names = ["월", "화", "수", "목", "금", "토", "일"]
print("■ 요일별 행 수")
for i, c in pd.Series(df_raw.index.dayofweek).value_counts().sort_index().items():
    mark = "  ← 서울 휴장" if i >= 5 else ""
    print(f"   {names[i]}: {c}{mark}")

print(f"\n■ 주말 행       : {int(is_weekend.sum())}개")
print(f"■ 한국 공휴일 행 : {int(is_holiday.sum())}개")
if is_holiday.any():
    print("   (예시)")
    for d in df_raw.index[is_holiday][:5]:
        print(f"     {d.date()}  {df_raw.loc[d, 'value']:,.2f}")

# 서울 거래일 기준 완전성: 열렸어야 하는 날 중 값이 있는 비율
biz = pd.bdate_range(df_raw.index.min(), df_raw.index.max())
expected = pd.DatetimeIndex([d for d in biz if d.date() not in KR_HOLIDAYS])
missing = expected.difference(df_raw.index)

print(f"\n■ 서울 거래일 기준")
print(f"   열렸어야 하는 날 : {len(expected)}일")
print(f"   데이터에 있는 날 : {len(expected) - len(missing)}일")
print(f"   진짜 결측       : {len(missing)}일")


def easter_sunday(y):
    # 부활절(춘분 이후 첫 보름 다음 일요일) 계산 — 해마다 날짜가 바뀐다
    a, b, c = y % 19, y // 100, y % 100
    d, e = b // 4, b % 4
    f, g = (b + 8) // 25, (b - (b + 8) // 25 + 1) // 3
    h = (19 * a + b - d - g + 15) % 30
    i, k = c // 4, c % 4
    l = (32 + 2 * e + 2 * i - h - k) % 7
    m = (a + 11 * h + 22 * l) // 451
    return date(y, (h + l - 7 * m + 114) // 31, ((h + l - 7 * m + 114) % 31) + 1)


def why_missing(d):
    # 서울은 열렸는데 데이터가 없는 날의 정체를 추정
    es = easter_sunday(d.year)
    if d.date() == es - timedelta(days=2):
        return "성금요일 (글로벌 외환시장 휴장, 한국은 개장)"
    if d.date() == es + timedelta(days=1):
        return "부활절 월요일 (글로벌 외환시장 휴장, 한국은 개장)"
    if (d.month, d.day) in [(12, 25), (1, 1)]:
        return "글로벌 공휴일"
    return "원인 미상 — 원본에서 직접 확인 필요"


if len(missing):
    print("\n   결측일 상세:")
    for d in missing:
        print(f"     {d.date()} ({names[d.dayofweek]}) — {why_missing(d)}")
    print("\n   💡 Yahoo의 KRW=X는 글로벌 장외 호가라 서양권 휴장일에는 값이 없습니다.")
    print("      한국 공휴일에는 값이 있고(불필요), 서양 휴일에는 값이 없는(부족) 구조입니다.")
````

````python
# ✏️ 한국 시장(환율·국내주식) 데이터면 True. 해외 지표·기온 등이면 False.
ALIGN_TO_KR_TRADING_DAYS = True

# ✏️ 서울은 열렸는데 원본에 값이 없는 날(성금요일 등)을 거래일 축에 추가할지
#    True  → 거래일 완전성 100%. 단 그날 변동률은 0%가 되어 실제 움직임이 아님
#    False → 구멍을 그대로 둠. 다음 거래일 변동률이 그 기간을 합산해 반영
FILL_MISSING_TRADING_DAYS = True

# ▶️ 아래는 실행 — 서울 거래일 기준으로 데이터를 정렬
TRADING_DAY_LOG = []

if ALIGN_TO_KR_TRADING_DAYS:
    drop = (is_weekend | is_holiday).values
    n_we, n_ho = int(is_weekend.sum()), int(is_holiday.sum())
    before = len(df_raw)
    df_raw = df_raw[~drop]
    TRADING_DAY_LOG.append(
        f"서울 외환시장 휴장일 {n_we + n_ho}행 제거(주말 {n_we}, 공휴일·근로자의날·연말 {n_ho}). "
        f"{before}행 → {len(df_raw)}행. "
        f"Yahoo의 KRW=X는 글로벌 장외 호가라 국내 휴장일에도 값이 들어오므로, "
        f"국내 거래일 기준으로 분석하기 위해 정렬함")

    biz = pd.bdate_range(df_raw.index.min(), df_raw.index.max())
    # expected = pd.DatetimeIndex([d for d in biz if d.date() not in KR_HOLIDAYS])
    expected = pd.DatetimeIndex([d for d in biz if d.date() not in KR_HOLIDAYS],
                            name=df_raw.index.name)
    missing = expected.difference(df_raw.index)
    TRADING_DAY_LOG.append(
        f"정렬 후 서울 거래일 {len(expected)}일 중 {len(expected) - len(missing)}일 확보 "
        f"(완전성 {(1 - len(missing) / len(expected)) * 100:.1f}%, 결측 {len(missing)}일)")

    if len(missing) and FILL_MISSING_TRADING_DAYS:
        detail = ", ".join(f"{d.date()}({why_missing(d).split(' ')[0]})" for d in missing)
        df_raw = df_raw.reindex(expected)
        TRADING_DAY_LOG.append(
            f"서울은 개장했으나 원본에 값이 없는 {len(missing)}일을 거래일 축에 추가함: {detail}. "
            f"값은 STEP 3의 결측치 정책으로 채우며, 해당일 변동률은 실제 시장 움직임이 아님에 주의")
    elif len(missing):
        TRADING_DAY_LOG.append(
            f"서울 개장일 {len(missing)}일이 원본에 없어 그대로 비워 둠 "
            f"(전체의 {len(missing) / len(expected) * 100:.1f}%). "
            f"다음 거래일의 변동률이 그 기간을 합산해 반영한다는 점에 주의")
    TRADING_DAY_LOG.append(
        "종가 기준 차이: 서울은 15:30 주간종가, Yahoo는 야간거래를 포함한 하루 끝 호가. "
        "2025-04-09 기준 약 2원 차이가 확인되므로 절대 수준보다 변화율·추세 해석에 사용")
else:
    TRADING_DAY_LOG.append("거래일 정렬 미적용 (한국 시장 데이터가 아님)")

print("📅 거래일 정합성")
for i, log in enumerate(TRADING_DAY_LOG, 1):
    print(f"  {i}. {log}")
print(f"\n최종 데이터: {len(df_raw)}행 "
      f"({df_raw.index.min().date()} ~ {df_raw.index.max().date()})")
print(f"주말 행 {int((df_raw.index.dayofweek >= 5).sum())}개, "
      f"공휴일 행 {int(pd.Series(df_raw.index.date).isin(KR_HOLIDAYS).sum())}개 → 0이어야 정상")
````

|---|---|---|
| `ffill` | 직전 거래일 값으로 채움 | 환율처럼 "값이 유지된다"고 볼 수 있을 때 (기본 추천) |
| `interpolate` | 앞뒤 값으로 선형 보간 | 기온처럼 연속적으로 변하는 값 |
| `drop` | 결측 행 삭제 | 결측이 아주 적고 시점 연속성이 덜 중요할 때 |

### 이상치 탐지 방법 고르기

`OUTLIER_METHOD` 값만 바꾸면 방식이 교체되고, 선택한 기준이 리포트에 그대로 기록됩니다.

| 값 | 방법 | 판정 기준 | 특징 |
|---|---|---|---|
| `"iqr"` | **사분위수 범위 (기본)** | Q1−k×IQR 미만, Q3+k×IQR 초과 | 순위 기반이라 극단값에 덜 흔들림. 분포 가정 불필요 |
| `"zscore"` | **표준점수** | 평균에서 k×표준편차 이상 벗어남 | 계산이 직관적. 단 평균·표준편차 자체가 이상치에 끌려감 |
| `"hampel"` | **이동 중앙값 + MAD** | 최근 N일 중앙값에서 k×MAD 이상 벗어남 | 시간에 따라 기준이 움직임. 추세가 있는 시계열에 적합 |
| `"domain"` | **도메인 고정 임계값** | 일간 변동률 절대값이 X% 초과 | 근거를 도메인 지식으로 댈 수 있을 때 가장 설명하기 쉬움 |

⚠️ **탐지 ≠ 제거**: 시계열에서 급등·급락은 *에러가 아니라 분석 대상*인 경우가 많습니다.
`OUTLIER_POLICY` 기본값 `"keep"`(표시만 하고 유지)도 **정당한 근거가 있는 처리**이며, 이유를 리포트에 쓰면 됩니다.

````python
# ✏️ 결측치 처리 방법을 고르세요 ------------------------------------
MISSING_POLICY = "ffill"      # "ffill" | "interpolate" | "drop"
# ------------------------------------------------------------------

df = df_raw.copy()
CLEANING_LOG = list(TRADING_DAY_LOG)   # STEP 2-2의 거래일 정렬 기록을 이어받음

# 정제 전 상태를 기록해 두었다가 마지막에 전후를 비교한다
BEFORE = {
    "행 수": len(df),
    "날짜 범위": f"{df.index.min().date()} ~ {df.index.max().date()}",
    "결측 수": int(df["value"].isna().sum()),
    "평균": round(float(df["value"].mean()), 2),
}

# ── 1) 정렬 ────────────────────────────────────────────────────────
df = df.sort_index()
CLEANING_LOG.append("정렬: 날짜형으로 변환 후 시간 순서로 정렬")

# ── 2) 중복 ────────────────────────────────────────────────────────
dup = int(df.index.duplicated().sum())
if dup:
    df = df[~df.index.duplicated(keep="first")]
    CLEANING_LOG.append(f"중복 날짜 {dup}건 → 첫 번째 값만 유지")
else:
    CLEANING_LOG.append("중복 날짜 0건 → 별도 처리 없음")

# ── 3) 결측치 ──────────────────────────────────────────────────────
n_missing = int(df["value"].isna().sum())
if n_missing:
    if MISSING_POLICY == "ffill":
        df["value"] = df["value"].ffill().bfill()
        how = "직전 거래일 값으로 채움(ffill). 시장이 닫혀 있으면 값이 유지된다고 가정"
    elif MISSING_POLICY == "interpolate":
        df["value"] = df["value"].interpolate(method="linear").ffill().bfill()
        how = "선형 보간(interpolate). 두 관측 사이에서 일정한 속도로 변한다고 가정"
    else:
        df = df.dropna(subset=["value"])
        how = "결측 행 삭제(drop). 관측된 날만으로도 대표성이 있다고 판단"
    CLEANING_LOG.append(f"결측치 {n_missing}건 → {how}")
else:
    CLEANING_LOG.append("결측치 0건 → 별도 처리 없음")

# ── 4) 이상치: 기록 오류가 있는지만 점검한다 ────────────────────────
# 위 표의 탐지 방법들은 "기록 오류"를 찾는 도구입니다.
# 이 데이터는 시장 체결값을 그대로 받은 것이라 오류가 끼어들 여지가 거의 없습니다.
# 그래서 통계적 탐지를 적용하는 대신, 오류로 볼 값이 있는지만 확인합니다.
err = []
if (df["value"] <= 0).any():
    err.append(f"0 이하의 값 {int((df['value'] <= 0).sum())}건 (환율에 불가능)")
if df["value"].isna().any():
    err.append(f"미처리 결측 {int(df['value'].isna().sum())}건")

LO, HI = 500, 3000                                   # 자릿수 오류를 잡는 상식 범위
if ((df["value"] < LO) | (df["value"] > HI)).any():
    err.append(f"상식 범위({LO}~{HI}원) 밖 존재 — 자릿수 오류 의심")

impossible = int((df["value"].pct_change().abs() > 0.10).sum())
if impossible:
    err.append(f"일간 변동률 10% 초과 {impossible}건 — 기록 오류 가능성 확인 필요")

print("🔍 이상치 점검 — 기록 오류가 있는가")
if err:
    for e in err:
        print(f"   ❌ {e}")
    print("\n   → 오류로 의심되는 값이 있습니다. 원본을 확인한 뒤 처리 방법을 정하세요.")
    OUTLIER_LOG = "기록 오류 점검 결과 확인된 문제: " + " / ".join(err)
else:
    print("   ✅ 0 이하 값 없음")
    print("   ✅ 미처리 결측 없음")
    print(f"   ✅ 값이 모두 상식 범위({LO}~{HI}원) 안")
    print("   ✅ 일간 변동률 10% 초과 없음")
    print("\n   → 기록 오류로 볼 값이 없습니다. 이 데이터셋에는 제거할 이상치가 없습니다.")
    print("     크게 움직인 날들은 오류가 아니라 실제 시장 움직임이므로 그대로 둡니다.")
    OUTLIER_LOG = ("이상치: 0 이하 값·미처리 결측·상식 범위·일간 10% 초과를 점검한 결과 "
                   "기록 오류에 해당하는 값이 없어 제거·수정 없이 전부 유지. "
                   "급등·급락은 측정 오류가 아니라 실제 시장 움직임으로 판단")
CLEANING_LOG.append(OUTLIER_LOG)

# ── 정제 전후 비교 ─────────────────────────────────────────────────
AFTER = {
    "행 수": len(df),
    "날짜 범위": f"{df.index.min().date()} ~ {df.index.max().date()}",
    "결측 수": int(df["value"].isna().sum()),
    "평균": round(float(df["value"].mean()), 2),
}

print("\n🧹 정제 로그")
for i, log in enumerate(CLEANING_LOG, 1):
    print(f"  {i}. {log}")

print("\n📊 정제 전후 비교")
display(pd.DataFrame({"정제 전": BEFORE, "정제 후": AFTER}))
print(f"최종 데이터: {len(df)}행")
````

````python
# ▶️ 실행 — STEP 3 보완: 이상치 탐지 설정을 변수로 남긴다

OUTLIER_METHOD = "domain"   # "iqr" | "zscore" | "hampel" | "domain"
OUTLIER_POLICY = "keep"     # "keep"(표시만 하고 유지) | "clip"(임계값으로 눌러 담기)

DOMAIN_LIMIT = 10.0         # 일간 변동률 절대값 10% 초과 = 기록 오류 의심 (위 점검과 같은 기준)
IQR_K        = 1.5          # 아래 세 줄은 방법을 바꿀 때를 대비한 기본값
Z_THRESHOLD  = 3.0
HAMPEL_WIN, HAMPEL_K = 21, 3.0

df["is_outlier"] = df["value"].pct_change().abs().fillna(0) > DOMAIN_LIMIT / 100
print(f"이상치 탐지={OUTLIER_METHOD} (임계 {DOMAIN_LIMIT}%) / 정책={OUTLIER_POLICY} "
      f"→ 표시된 값 {int(df['is_outlier'].sum())}건 (제거·수정 없음)")
````

# STEP 4. 시계열 시각화 분석 기법 적용 ▶️

미션 요구: **기법 최소 2가지** → 여기서 4가지를 한 번에 계산합니다.

| 기법 | 무엇을 보나 | 왜 쓰나 |
|---|---|---|
| **이동평균 (MA)** | 노이즈 걷어낸 **추세** | 하루하루 흔들림 때문에 안 보이던 방향을 드러냄 |
| **변화율 (일간 변동률)** | 하루 사이 **변동 폭** | 환율 수준(1,300원)보다 "하루에 몇 % 움직였나"가 중요할 때 |
| **롤링 변동성 (표준편차)** | 외환시장이 **불안정한 구간** | 같은 상승도 완만한 상승 vs 요동친 상승은 의미가 다름 |
| **월별 집계** | **계절성 / 주기** | 특정 월에 반복되는 패턴이 있는지 |

> 📌 **면접·발표 대비 한 줄 정의**
> - **추세(Trend)**: 장기적으로 올라가거나 내려가는 방향
> - **계절성(Seasonality)**: 일정 주기로 반복되는 패턴 (월별·요일별 등)
> - **노이즈(Noise)**: 추세·계절성으로 설명 안 되는 불규칙한 잔여 변동

````python
# ✏️ 이동평균 창 크기 (일 단위). 데이터 성격에 맞게 바꿔도 됩니다.
MA_SHORT, MA_LONG, VOL_WINDOW = 7, 30, 30

# ▶️ 아래는 실행
df["ma_short"]   = df["value"].rolling(MA_SHORT).mean()
df["ma_long"]    = df["value"].rolling(MA_LONG).mean()
df["ret"]        = df["value"].pct_change() * 100          # 일간 변동률(%) — 양수면 원화 약세
df["volatility"] = df["ret"].rolling(VOL_WINDOW).std()     # 롤링 변동성
df["cum_ret"]    = (df["value"] / df["value"].iloc[0] - 1) * 100  # 시작 대비 누적 변동률(%)

# 월별 집계 (pandas 버전에 따라 "ME"/"M" 자동 선택)
try:
    monthly = df["value"].resample("ME").last()
except ValueError:
    monthly = df["value"].resample("M").last()
monthly_ret = (monthly.pct_change() * 100).dropna()
month_avg = monthly_ret.groupby(monthly_ret.index.month).mean()

# 요일별 집계
weekday_avg = df.groupby(df.index.dayofweek)["ret"].mean()
weekday_names = ["월", "화", "수", "목", "금", "토", "일"]

print(f"✅ 적용한 기법 4가지: 이동평균(MA{MA_SHORT}/MA{MA_LONG}), 변화율, "
      f"롤링 변동성({VOL_WINDOW}일), 월별/요일별 집계")
display(df[["value", "ma_short", "ma_long", "ret", "volatility"]].tail())
````

# 🎁 STEP 6. 보너스 (A) 시계열 분해 ▶️

원본 시계열을 **추세 + 계절성 + 잔차(노이즈)** 로 분리합니다.
"올라간 게 진짜 추세인가, 아니면 계절적 반복인가"를 눈으로 구분할 수 있게 해줍니다.

````python
# ▶️ 실행만
from statsmodels.tsa.seasonal import seasonal_decompose

# ✏️ 주기: 환율은 주간 주기 5(영업일) 또는 월간 30을 주로 씁니다
DECOMP_PERIOD = 30

# 분해 함수는 간격이 일정한 시계열을 요구한다. 거래일 데이터에는 주말·휴장일 구멍이 있으므로
# 달력일로 채운 뒤 보간한다. 보간된 값은 실제 거래가 아니라는 점을 리포트에 밝힐 것.
s = df["value"].asfreq("D").interpolate().ffill().bfill()
result = seasonal_decompose(s, model="additive", period=DECOMP_PERIOD)

fig, axes = plt.subplots(4, 1, figsize=(12, 9), sharex=True)
for ax, data, name, c in zip(
    axes,
    [result.observed, result.trend, result.seasonal, result.resid],
    ["원본 (Observed)", "추세 (Trend)", f"계절성 (Seasonality, 주기 {DECOMP_PERIOD}일)", "잔차 (Residual)"],
    ["#34495E", "#E74C3C", "#2E86DE", "#95A5A6"],
):
    ax.plot(data.index, data.values, color=c, lw=1.4)
    ax.set_title(name, fontsize=12, loc="left")
axes[-1].set_xlabel("날짜")

save_fig(fig, "07_decomposition.png", "시계열 분해 (추세·계절성·잔차)",
         f"가법 모형(additive) 기준 시계열 분해. 주기 {DECOMP_PERIOD}일로 설정했다. 분해 함수가 균등 간격을 요구하므로 주말·휴장일은 보간으로 채웠고, 따라서 계절성 성분에는 실제 거래가 없던 날의 값이 섞여 있다. 잔차가 특정 구간에서만 크게 튄다면 그 시점에 외부 충격이 있었을 가능성이 있다.")

resid_std = float(result.resid.dropna().std())
trend_change = float(result.trend.dropna().iloc[-1] - result.trend.dropna().iloc[0])
print(f"\n추세 성분 총 변화량 : {trend_change:,.2f}")
print(f"잔차 표준편차      : {resid_std:,.2f}")
````

# 🎁 STEP 7. 보너스 (B) 간단 예측 (베이스라인) ▶️

미션 안내대로 **정확도보다 "가정과 한계" 설명**이 핵심입니다.

| 베이스라인 | 가정 |
|---|---|
| **Naive** | "내일은 오늘과 같다" |
| **이동평균** | "최근 N일 평균 수준이 이어진다" |
| **Drift** | "최근 평균 변화 속도가 그대로 이어진다" |

이보다 못한 예측 모델은 쓸 이유가 없다 — 그래서 **베이스라인이 기준선**입니다.

````python
# ✏️ 검증 구간 길이 (마지막 N일을 테스트로 씁니다)
TEST_DAYS = 30

train, test = df["value"].iloc[:-TEST_DAYS], df["value"].iloc[-TEST_DAYS:]

pred_naive = np.repeat(train.iloc[-1], len(test))
pred_ma    = np.repeat(train.iloc[-MA_LONG:].mean(), len(test))
slope      = (train.iloc[-1] - train.iloc[0]) / (len(train) - 1)
pred_drift = train.iloc[-1] + slope * np.arange(1, len(test) + 1)

def mae(a, p):  return float(np.mean(np.abs(a - p)))
def mape(a, p): return float(np.mean(np.abs((a - p) / a)) * 100)

scores = {}
for name, pred in [("Naive(마지막 값)", pred_naive),
                   (f"이동평균({MA_LONG}일)", pred_ma),
                   ("Drift(추세 연장)", pred_drift)]:
    scores[name] = {"MAE": mae(test.values, pred), "MAPE(%)": mape(test.values, pred)}

score_df = pd.DataFrame(scores).T.round(3)
display(score_df)
best = score_df["MAPE(%)"].idxmin()
print(f"🏆 가장 오차가 작은 베이스라인: {best} (MAPE {score_df.loc[best,'MAPE(%)']:.2f}%)")

fig, ax = plt.subplots()
ax.plot(train.index[-90:], train.iloc[-90:], color="#34495E", lw=1.6, label="학습 구간(최근 90일)")
ax.plot(test.index, test.values, color="#16A085", lw=2.2, label="실제값(검증 구간)")
ax.plot(test.index, pred_naive, "--", lw=1.8, label="Naive")
ax.plot(test.index, pred_ma, "--", lw=1.8, label=f"이동평균 {MA_LONG}일")
ax.plot(test.index, pred_drift, "--", lw=1.8, label="Drift")
ax.axvline(test.index[0], color="gray", ls=":", lw=1.2)
ax.set_title(f"베이스라인 예측 vs 실제 (마지막 {TEST_DAYS}일)", fontsize=14, pad=12)
ax.set_xlabel("날짜"); ax.set_ylabel(VALUE_LABEL); ax.legend()

save_fig(fig, "08_baseline_forecast.png", "베이스라인 예측 vs 실제",
         f"마지막 {TEST_DAYS}일을 검증 구간으로 떼어내 세 가지 베이스라인을 비교했다. 외부 변수를 전혀 쓰지 않은 단순 모형이므로 실제 예측용이 아니라 '기준선'으로만 해석해야 한다.")

FORECAST_NOTE = (f"검증 구간 {TEST_DAYS}일 기준 최소 오차 모델은 {best} "
                 f"(MAPE {score_df.loc[best,'MAPE(%)']:.2f}%). "
                 "과거 환율 값만 사용했고 외부 변수(금리, 물가지표, 달러인덱스 등)를 반영하지 않았다는 가정 위에 있다.")
print("\n" + FORECAST_NOTE)
````

# 🎁 STEP 8. 보너스 (C) 인터랙티브 대시보드 ▶️

기간·이동평균 창을 슬라이더로 바꾸며 탐색합니다.
제출 시에는 **이 화면 스크린샷 + 필터 변경 시나리오 설명**으로 (3)번 옵션을 충족할 수 있습니다.

````python
# ▶️ 실행 후 슬라이더를 움직여 보세요
import ipywidgets as widgets
from ipywidgets import interact

dates = df.index
opts = [(d.strftime("%Y-%m-%d"), i) for i, d in enumerate(dates)]

@interact(
    구간=widgets.SelectionRangeSlider(options=opts, index=(0, len(opts) - 1),
                                     description="기간", continuous_update=False,
                                     layout=widgets.Layout(width="700px")),
    이동평균창=widgets.IntSlider(value=MA_LONG, min=3, max=90, step=1, description="MA 창"),
    변동성표시=widgets.Checkbox(value=True, description="변동성 함께 보기"),
)
def dashboard(구간, 이동평균창, 변동성표시):
    a, b = 구간
    sub = df.iloc[a:b + 1].copy()
    if len(sub) < 5:
        print("구간이 너무 짧습니다."); return
    sub["ma"] = sub["value"].rolling(이동평균창).mean()

    rows = 2 if 변동성표시 else 1
    fig, axes = plt.subplots(rows, 1, figsize=(11, 6 if 변동성표시 else 4), sharex=True)
    axes = np.atleast_1d(axes)
    axes[0].plot(sub.index, sub["value"], lw=1, alpha=0.5, color="#8899aa", label="원본")
    axes[0].plot(sub.index, sub["ma"], lw=2.2, color="#E74C3C", label=f"MA{이동평균창}")
    axes[0].legend(); axes[0].set_ylabel(VALUE_LABEL)
    if 변동성표시:
        axes[1].plot(sub.index, sub["ret"].rolling(이동평균창).std(), color="#8E44AD", lw=1.6)
        axes[1].set_ylabel("변동성 (%)")
    fig.suptitle(f"{sub.index[0].date()} ~ {sub.index[-1].date()}", fontsize=13)
    plt.tight_layout(); plt.show()

    chg = (sub["value"].iloc[-1] / sub["value"].iloc[0] - 1) * 100
    print(f"📊 선택 구간 {len(sub)}일 | 변화율 {chg:+.2f}% | "
          f"최고 {sub['value'].max():,.2f} | 최저 {sub['value'].min():,.2f} | "
          f"일간 변동성 {sub['ret'].std():.2f}%")
````

````python
# ▶️ 실행
FACTS = {}
FACTS["기간"] = f"{df.index.min().date()} ~ {df.index.max().date()}"
FACTS["데이터 수"] = f"{len(df)}개"
FACTS["시작값"] = f"{df['value'].iloc[0]:,.2f}"
FACTS["종료값"] = f"{df['value'].iloc[-1]:,.2f}"
FACTS["전체 변화율"] = f"{(df['value'].iloc[-1]/df['value'].iloc[0]-1)*100:+.2f}%"
FACTS["최고점"] = f"{df['value'].max():,.2f} ({df['value'].idxmax().date()})"
FACTS["최저점"] = f"{df['value'].min():,.2f} ({df['value'].idxmin().date()})"
FACTS["일간 최대 상승(원화 약세)"] = f"{df['ret'].max():+.2f}% ({df['ret'].idxmax().date()})"
FACTS["일간 최대 하락(원화 강세)"] = f"{df['ret'].min():+.2f}% ({df['ret'].idxmin().date()})"
FACTS["일간 변동성(표준편차)"] = f"{df['ret'].std():.2f}%"
FACTS["환율이 오른 날 비율"] = f"{(df['ret'] > 0).mean()*100:.1f}%"

_bm, _wm = month_avg.idxmax(), month_avg.idxmin()
FACTS["평균 변동률 최고 월"] = f"{_bm}월 ({month_avg.max():+.2f}%)"
FACTS["평균 변동률 최저 월"] = f"{_wm}월 ({month_avg.min():+.2f}%)"

_vol = df["volatility"].dropna()
if len(_vol):
    FACTS["변동성 최고 시점"] = f"{_vol.max():.2f}% ({_vol.idxmax().date()})"

# 전반부 vs 후반부 비교 (추세가 구간별로 다른지)
half = len(df) // 2
f_chg = (df["value"].iloc[half-1]/df["value"].iloc[0]-1)*100
s_chg = (df["value"].iloc[-1]/df["value"].iloc[half]-1)*100
FACTS["전반부 변화율"] = f"{f_chg:+.2f}%"
FACTS["후반부 변화율"] = f"{s_chg:+.2f}%"

# 이동평균 교차 (단기선이 장기선을 돌파한 지점 = 추세 전환 후보)
cross = np.sign(df["ma_short"] - df["ma_long"]).diff()
up_cross   = df.index[cross == 2]    # 단기선이 위로 돌파 → 원화 약세 전환 후보
down_cross = df.index[cross == -2]   # 단기선이 아래로 돌파 → 원화 강세 전환 후보
FACTS["이동평균 상향 교차(약세 전환 후보)"] = f"{len(up_cross)}회"
FACTS["이동평균 하향 교차(강세 전환 후보)"] = f"{len(down_cross)}회"
if len(up_cross):   FACTS["최근 상향 교차"] = str(up_cross[-1].date())
if len(down_cross): FACTS["최근 하향 교차"] = str(down_cross[-1].date())

print("=" * 58)
print("📌 인사이트에 쓸 수 있는 관찰(Fact) 목록")
print("=" * 58)
for k, v in FACTS.items():
    print(f"  {k:<22}: {v}")
print("=" * 58)
print("\n👉 이 숫자들 중 3개 이상을 골라 STEP 10에서 '관찰'로 쓰고,")
print("   '왜 그런가(해석)'는 반드시 본인이 채우세요.")
````

**이 노트북 파일 자체는 제출물이 아닙니다.** 여기에는 STEP 안내, TODO, 체크리스트 같은 학습용 장치가 가득해서, 그대로 올리면 동료평가자가 "분석"이 아니라 "실습 교재"를 읽게 됩니다.

그래서 **당신이 고른 설정 그대로 돌아가는 깔끔한 스크립트**를 따로 만들어 저장합니다. 이게 저장소에 올라갈 코드입니다.

````python
# ▶️ 실행 — 제출용 analysis.py 생성
header = f'''"""{TOPIC_TITLE}

데이터 출처 : {DATA_SOURCE}
분석 기간   : {FACTS["기간"]}
실행 방법   : python analysis.py   (images/ 폴더에 그래프가 생성됩니다)
"""
import os
from datetime import date

import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt

# ---- 설정 ----------------------------------------------------------
TICKER, START_DATE, END_DATE = "{TICKER}", "{START_DATE}", "{END_DATE}"
DATA_PATH   = "data/{PROJECT_NAME}.csv"
IMG_DIR     = "images"
TOPIC_TITLE = "{TOPIC_TITLE}"
VALUE_LABEL = "{VALUE_LABEL}"

ALIGN_TO_KR_TRADING_DAYS = {ALIGN_TO_KR_TRADING_DAYS}
MISSING_POLICY = "{MISSING_POLICY}"
OUTLIER_METHOD = "{OUTLIER_METHOD}"
OUTLIER_POLICY = "{OUTLIER_POLICY}"
IQR_K, Z_THRESHOLD = {IQR_K}, {Z_THRESHOLD}
HAMPEL_WIN, HAMPEL_K, DOMAIN_LIMIT = {HAMPEL_WIN}, {HAMPEL_K}, {DOMAIN_LIMIT}
MA_SHORT, MA_LONG, VOL_WINDOW = {MA_SHORT}, {MA_LONG}, {VOL_WINDOW}
# --------------------------------------------------------------------
'''

body = '''
def setup_korean_font():
    """한글 폰트를 등록한다. 폰트 파일을 직접 addfont 해야 새로 설치한 폰트도 인식된다."""
    import matplotlib.font_manager as fm

    candidates = [
        "/usr/share/fonts/truetype/nanum/NanumGothic.ttf",       # Linux (apt fonts-nanum)
        "/usr/share/fonts/truetype/nanum/NanumBarunGothic.ttf",
        "/System/Library/Fonts/AppleSDGothicNeo.ttc",            # macOS
        "/Library/Fonts/AppleGothic.ttf",
        "C:/Windows/Fonts/malgun.ttf",                           # Windows
    ]
    candidates += [f for f in fm.findSystemFonts()
                   if any(k in os.path.basename(f) for k in ("Nanum", "malgun", "AppleGothic"))]

    for path in candidates:
        if os.path.exists(path):
            try:
                fm.fontManager.addfont(path)
                name = fm.FontProperties(fname=path).get_name()
                plt.rcParams["font.family"] = name
                return name
            except Exception:
                continue
    return None


_font = setup_korean_font()
print(f"[폰트] 한글 폰트: {_font or '없음 — 그래프의 한글이 깨질 수 있습니다'}")
plt.rcParams["axes.unicode_minus"] = False
plt.rcParams["figure.figsize"] = (12, 5)
plt.rcParams["axes.grid"] = True
plt.rcParams["grid.alpha"] = 0.3
os.makedirs(IMG_DIR, exist_ok=True)


def load_data():
    """저장된 CSV 스냅샷을 우선 사용하고, 없으면 원본에서 내려받는다."""
    if os.path.exists(DATA_PATH):
        d = pd.read_csv(DATA_PATH, parse_dates=["date"]).set_index("date")
        return d[["value"]].sort_index()
    import yfinance as yf
    raw = yf.download(TICKER, start=START_DATE, end=END_DATE,
                      progress=False, auto_adjust=True)
    if isinstance(raw.columns, pd.MultiIndex):
        raw.columns = raw.columns.get_level_values(0)
    d = pd.DataFrame({"value": raw["Close"].astype(float)})
    d.index.name = "date"
    os.makedirs(os.path.dirname(DATA_PATH), exist_ok=True)
    d.to_csv(DATA_PATH, encoding="utf-8-sig")
    return d.sort_index()


def kr_market_holidays(years):
    """공휴일 + 근로자의날(5/1) + 연말 휴장일(12/31). 금융시장은 이 날 모두 쉰다."""
    import holidays as pyholidays
    hs = set(pyholidays.KR(years=list(years)).keys())
    for y in years:
        hs.add(date(y, 5, 1))
        hs.add(date(y, 12, 31))
    return hs


def align_to_kr_trading_days(d):
    """Yahoo의 KRW=X는 글로벌 장외 호가라 국내 휴장일에도 값이 있다.

    서울 외환시장 거래일만 남겨 국내 기준으로 분석할 수 있게 정렬한다.
    """
    if not ALIGN_TO_KR_TRADING_DAYS:
        return d
    try:
        hol = kr_market_holidays(range(d.index.min().year, d.index.max().year + 1))
    except ImportError:
        print("[거래일] holidays 패키지가 없어 주말만 제거합니다 (pip install holidays)")
        hol = set()
    keep = (d.index.dayofweek < 5) & (~pd.Series(d.index.date, index=d.index).isin(hol))
    before = len(d)
    d = d[keep.values]
    print(f"[거래일] 서울 휴장일 {before - len(d)}행 제거 → {len(d)}행")
    return d


def clean(d):
    """결측치를 처리하고, 선택한 방법으로 이상치를 탐지한다."""
    d = d[~d.index.duplicated(keep="first")].copy()
    if MISSING_POLICY == "ffill":
        d["value"] = d["value"].ffill().bfill()
    elif MISSING_POLICY == "interpolate":
        d["value"] = d["value"].interpolate("linear").ffill().bfill()
    else:
        d = d.dropna(subset=["value"])

    ret = d["value"].pct_change()
    low = high = None
    if OUTLIER_METHOD == "iqr":
        q1, q3 = ret.quantile(0.25), ret.quantile(0.75)
        low, high = q1 - IQR_K * (q3 - q1), q3 + IQR_K * (q3 - q1)
        mask = (ret < low) | (ret > high)
    elif OUTLIER_METHOD == "zscore":
        low = ret.mean() - Z_THRESHOLD * ret.std()
        high = ret.mean() + Z_THRESHOLD * ret.std()
        mask = ((ret - ret.mean()) / ret.std()).abs() > Z_THRESHOLD
    elif OUTLIER_METHOD == "hampel":
        med = ret.rolling(HAMPEL_WIN, center=True, min_periods=5).median()
        mad = (ret - med).abs().rolling(HAMPEL_WIN, center=True, min_periods=5).median()
        mask = (ret - med).abs() > HAMPEL_K * 1.4826 * mad
    else:
        low, high = -DOMAIN_LIMIT / 100, DOMAIN_LIMIT / 100
        mask = ret.abs() > DOMAIN_LIMIT / 100

    mask = mask.fillna(False)
    if OUTLIER_POLICY == "clip" and mask.any() and low is not None:
        base = d["value"].iloc[0]
        d["value"] = base * (1 + ret.clip(low, high).fillna(0)).cumprod()
    d["is_outlier"] = mask
    print(f"[정제] 결측={MISSING_POLICY} / 이상치 탐지={OUTLIER_METHOD} "
          f"{int(mask.sum())}건 (정책={OUTLIER_POLICY})")
    return d


def add_features(d):
    """이동평균, 일간 변동률, 롤링 변동성, 누적 변동률을 계산한다."""
    d["ma_short"] = d["value"].rolling(MA_SHORT).mean()
    d["ma_long"] = d["value"].rolling(MA_LONG).mean()
    d["diff"] = d["value"].diff()
    d["ret"] = d["value"].pct_change() * 100
    d["volatility"] = d["ret"].rolling(VOL_WINDOW).std()
    d["cum_ret"] = (d["value"] / d["value"].iloc[0] - 1) * 100
    return d


def plot_all(d):
    """리포트에 사용한 그래프 4개를 images/ 에 저장한다."""
    try:
        monthly = d["value"].resample("ME").last()
    except ValueError:
        monthly = d["value"].resample("M").last()
    monthly_ret = (monthly.pct_change() * 100).dropna()
    month_avg = monthly_ret.groupby(monthly_ret.index.month).mean()

    fig, ax = plt.subplots()
    ax.plot(d.index, d["value"], lw=1, alpha=0.45, color="#8899aa", label="원본")
    ax.plot(d.index, d["ma_short"], lw=1.8, color="#2E86DE", label=f"{MA_SHORT}일 이동평균")
    ax.plot(d.index, d["ma_long"], lw=2.2, color="#E74C3C", label=f"{MA_LONG}일 이동평균")
    ax.set_title(f"{TOPIC_TITLE} — 전체 추세와 이동평균", fontsize=14, pad=12)
    ax.set_xlabel("날짜"); ax.set_ylabel(VALUE_LABEL); ax.legend()
    fig.tight_layout(); fig.savefig(f"{IMG_DIR}/01_trend_moving_average.png", dpi=130)
    plt.close(fig)

    fig, axes = plt.subplots(2, 1, figsize=(12, 7), sharex=True)
    axes[0].bar(d.index, d["ret"], width=1.0,
                color=np.where(d["ret"] >= 0, "#E74C3C", "#2E86DE"), alpha=0.75)
    axes[0].axhline(0, color="black", lw=0.8)
    axes[0].set_title("일간 변동률 (%)  ▲ 환율 상승 = 원화 약세", fontsize=13)
    axes[1].plot(d.index, d["volatility"], color="#8E44AD", lw=1.8)
    axes[1].fill_between(d.index, d["volatility"], alpha=0.2, color="#8E44AD")
    axes[1].axhline(d["volatility"].mean(), ls="--", color="gray", lw=1,
                    label=f"평균 {d['volatility'].mean():.2f}%")
    axes[1].set_title(f"{VOL_WINDOW}일 롤링 변동성", fontsize=13)
    axes[1].set_xlabel("날짜"); axes[1].legend()
    fig.tight_layout(); fig.savefig(f"{IMG_DIR}/02_return_volatility.png", dpi=130)
    plt.close(fig)

    fig, axes = plt.subplots(1, 2, figsize=(13, 4.6))
    axes[0].bar(range(len(monthly_ret)), monthly_ret.values,
                color=["#E74C3C" if v >= 0 else "#2E86DE" for v in monthly_ret.values],
                alpha=0.85)
    axes[0].set_xticks(range(len(monthly_ret)))
    axes[0].set_xticklabels([x.strftime("%y-%m") for x in monthly_ret.index],
                            rotation=90, fontsize=8)
    axes[0].axhline(0, color="black", lw=0.8)
    axes[0].set_title("월별 환율 변동률 추이 (%)", fontsize=13)
    axes[1].bar([f"{m}월" for m in month_avg.index], month_avg.values,
                color=["#E74C3C" if v >= 0 else "#2E86DE" for v in month_avg.values],
                alpha=0.85)
    axes[1].axhline(0, color="black", lw=0.8)
    axes[1].set_title("월(1~12)별 평균 변동률 (%)", fontsize=13)
    fig.tight_layout(); fig.savefig(f"{IMG_DIR}/03_monthly_return.png", dpi=130)
    plt.close(fig)

    weekday_names = ["월", "화", "수", "목", "금", "토", "일"]
    wd = d.groupby(d.index.dayofweek)["ret"].mean()
    wd = wd[wd.index <= 4]
    fig, axes = plt.subplots(1, 2, figsize=(13, 4.6))
    axes[0].plot(d.index, d["cum_ret"], color="#16A085", lw=2)
    axes[0].fill_between(d.index, d["cum_ret"], alpha=0.18, color="#16A085")
    axes[0].axhline(0, color="black", lw=0.8)
    axes[0].set_title("시작 시점 대비 누적 변동률 (%)  ▲ 원화 약세", fontsize=13)
    axes[1].bar([weekday_names[i] for i in wd.index], wd.values,
                color=["#E74C3C" if v >= 0 else "#2E86DE" for v in wd.values], alpha=0.85)
    axes[1].axhline(0, color="black", lw=0.8)
    axes[1].set_title("요일별 평균 변동률 (%)", fontsize=13)
    fig.tight_layout(); fig.savefig(f"{IMG_DIR}/04_cumulative_weekday.png", dpi=130)
    plt.close(fig)

    # 05 히스토그램 + 박스플롯
    fig, axes = plt.subplots(1, 2, figsize=(13, 4.6))
    r = d["ret"].dropna()
    axes[0].hist(r, bins=30, color="#2E86DE", alpha=0.8, edgecolor="white")
    axes[0].axvline(0, color="black", lw=0.9)
    axes[0].axvline(r.mean(), color="#E74C3C", ls="--", lw=1.4,
                    label=f"평균 {r.mean():.3f}%")
    axes[0].set_title("일간 변동률 분포 (히스토그램)", fontsize=13)
    axes[0].set_xlabel("일간 변동률 (%)"); axes[0].set_ylabel("일수"); axes[0].legend()

    years = sorted(d.index.year.unique())
    groups = [d.loc[d.index.year == y, "ret"].dropna().values for y in years]
    bp = axes[1].boxplot(groups, labels=[f"{y}년" for y in years],
                         patch_artist=True, whis=1.5)
    for box in bp["boxes"]:
        box.set(facecolor="#AED6F1", alpha=0.8)
    for med in bp["medians"]:
        med.set(color="#1B4F72", linewidth=2)
    axes[1].axhline(0, color="black", lw=0.9)
    axes[1].set_title("연도별 일간 변동률 (박스플롯, 수염 1.5xIQR)", fontsize=13)
    fig.tight_layout(); fig.savefig(f"{IMG_DIR}/05_distribution.png", dpi=130)
    plt.close(fig)

    # 06 월 x 요일 히트맵
    pv = (d.assign(월=d.index.month, 요일=d.index.dayofweek)
            .pivot_table(index="월", columns="요일", values="ret", aggfunc="mean"))
    pv = pv[[c for c in pv.columns if c <= 4]]
    fig, ax = plt.subplots(figsize=(9, 6))
    lim = np.nanmax(np.abs(pv.values))
    im = ax.imshow(pv.values, cmap="RdBu_r", aspect="auto", vmin=-lim, vmax=lim)
    ax.set_xticks(range(len(pv.columns)))
    ax.set_xticklabels([weekday_names[c] for c in pv.columns])
    ax.set_yticks(range(len(pv.index)))
    ax.set_yticklabels([f"{m}월" for m in pv.index])
    for a_ in range(pv.shape[0]):
        for b_ in range(pv.shape[1]):
            val = pv.values[a_, b_]
            if not np.isnan(val):
                ax.text(b_, a_, f"{val:+.2f}", ha="center", va="center", fontsize=9,
                        color="white" if abs(val) > lim * 0.6 else "black")
    ax.set_title("월 x 요일 평균 일간 변동률 (%)", fontsize=13, pad=12)
    fig.colorbar(im, ax=ax, label="평균 변동률 (%)")
    fig.tight_layout(); fig.savefig(f"{IMG_DIR}/06_heatmap.png", dpi=130)
    plt.close(fig)

    print("[시각화] images/ 에 그래프 6개 저장 완료")


def summarize(d):
    """리포트 인사이트의 근거가 된 핵심 수치를 출력한다."""
    print("-" * 52)
    print(f"기간        : {d.index.min().date()} ~ {d.index.max().date()} ({len(d)}개)")
    print(f"전체 변화율   : {(d['value'].iloc[-1] / d['value'].iloc[0] - 1) * 100:+.2f}%")
    print(f"최고 / 최저   : {d['value'].max():,.2f} / {d['value'].min():,.2f}")
    print(f"일간 변동성   : {d['ret'].std():.2f}%")
    print(f"평균 / 중앙값 : {d['value'].mean():,.2f} / {d['value'].median():,.2f}")
    q1, q3 = d["value"].quantile(0.25), d["value"].quantile(0.75)
    print(f"Q1 / Q3 / IQR : {q1:,.2f} / {q3:,.2f} / {q3 - q1:,.2f}")
    print("-" * 52)


if __name__ == "__main__":
    data = add_features(clean(align_to_kr_trading_days(load_data())))
    plot_all(data)
    summarize(data)
'''

with open(f"{BASE}/analysis.py", "w", encoding="utf-8") as f:
    f.write(header + body)

print("✅ analysis.py 생성 완료 — 저장소에 올라갈 코드는 이 파일입니다.")
print(f"   경로: {BASE}/analysis.py")
````
